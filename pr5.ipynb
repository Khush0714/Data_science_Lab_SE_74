```python
# Step 1: Import Libraries
import pandas as pd
import numpy as np

# Step 2: Load Dataset
df = pd.read_csv("College.csv")

# Step 3: Display Data
print("First 5 Rows:\n", df.head())

print("\nDataset Info:\n")
print(df.info())

# -----------------------------------
# Step 4: Filtering Data
# -----------------------------------

# Filter colleges with Apps > 10000
filtered_data = df[df['Apps'] > 10000]
print("\nColleges with Applications > 10000:\n", filtered_data.head())

# Filter Private colleges
private_colleges = df[df['Private'] == 'Yes']
print("\nPrivate Colleges:\n", private_colleges.head())

# -----------------------------------
# Step 5: Grouping Data
# -----------------------------------

grouped = df.groupby('Private')

# -----------------------------------
# Step 6: Aggregate Functions
# -----------------------------------

print("\nMean Values:\n", grouped.mean(numeric_only=True))

print("\nSum Values:\n", grouped.sum(numeric_only=True))

print("\nCount:\n", grouped.count())

print("\nMinimum Values:\n", grouped.min(numeric_only=True))

print("\nMaximum Values:\n", grouped.max(numeric_only=True))

# -----------------------------------
# Step 7: Specific Aggregation
# -----------------------------------

agg_result = grouped['Apps'].agg(['mean', 'min', 'max', 'count'])
print("\nApplications Statistics:\n", agg_result)

# -----------------------------------
# Step 8: Insights
# -----------------------------------

print("\nInsights:")
print("1. Private and public colleges differ in number of applications.")
print("2. Some colleges receive significantly higher applications.")
print("3. Grouping helps compare private vs public institutions.")
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Count of Private vs Public Colleges
plt.figure()
sns.countplot(x='Private', data=df)
plt.title("Private vs Public Colleges")
plt.xlabel("Type of College")
plt.ylabel("Count")
plt.show()

# 2. Applications Distribution
plt.figure()
sns.histplot(df['Apps'], bins=30, kde=True)
plt.title("Distribution of Applications")
plt.xlabel("Number of Applications")
plt.ylabel("Frequency")
plt.show()

# 3. Applications vs Acceptances
plt.figure()
sns.scatterplot(x='Apps', y='Accept', hue='Private', data=df)
plt.title("Applications vs Acceptances")
plt.xlabel("Applications")
plt.ylabel("Acceptances")
plt.show()

# 4. Boxplot (Apps by College Type)
plt.figure()
sns.boxplot(x='Private', y='Apps', data=df)
plt.title("Applications by College Type")
plt.xlabel("Private/Public")
plt.ylabel("Applications")
plt.show()
```
