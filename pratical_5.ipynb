```python
df = pd.read_csv("Student_performance_data _.csv", nrows=10000)
df.head()
```

```python
print("\nDataset Information:")
print(df.info())
```

```python
print("\nDataset Description:")
print(df.describe())
```

```python
high_score = df[df['GPA'] > 3.5]

print("\nHigh performing students (GPA > 3.5):")
print(high_score.head())
```

```python
import pandas as pd

df['Score_Category'] = pd.cut(
    df['GPA'],
    bins=[0, 2.5, 3.5, 4],
    labels=['Low', 'Medium', 'High']
)

print(df.head())
```

```python
grouped_data = df.groupby('Score_Category')

print("\nGrouped Data:")
print(grouped_data.head())
```

```python
# Group by Score_Category and calculate mean GPA
mean_scores = df.groupby('Score_Category')['GPA'].mean()

print("\nMean GPA per Score Category:")
print(mean_scores)
```

```python
# Count students in each category
category_counts = df['Score_Category'].value_counts()

print("\nNumber of students per Score Category:")
print(category_counts)
```

```python
# Min and Max GPA per Score_Category
score_stats = df.groupby('Score_Category')['GPA'].agg(['min', 'max'])

print("\nMinimum and Maximum GPA per Score Category:")
print(score_stats)
```
