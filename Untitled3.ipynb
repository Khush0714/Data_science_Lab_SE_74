```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

# Create a user-defined dataset
data = list(map(float ,input("enetr number sperated by space:").split()))
print("User-defined Dataset:", data)
```

```python
# Ensure data is a NumPy array for calculations
data_np = np.array(data)

# Compute mean using NumPy
mean_value = np.mean(data_np)

# Compute median using NumPy
median_value = np.median(data_np)

# Compute mode using SciPy (NumPy does not have a direct mode function)
mode_result = stats.mode(data_np)
mode_value = mode_result.mode # Handle scalar or array mode output

print(f"Mean: {mean_value:.2f}")
print(f"Median: {median_value:.2f}")
print(f"Mode: {mode_value}")
```
