```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('retail_sales_dataset (3).csv')

# Clean names
df.columns = df.columns.str.strip().str.replace(' ', '_')

# Convert date
df['Date'] = pd.to_datetime(df['Date'])
```

```python
top_products = df.groupby('Product_Category')['Total_Amount'].sum().sort_values(ascending=False).head(5)

top_products.plot(kind='bar', color='orange')
plt.title('Top 5 Product Categories by Sales')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

```python
sales_by_gender = df.groupby('Gender')['Total_Amount'].sum()

sales_by_gender.plot(kind='pie', autopct='%1.1f%%')
plt.title('Sales by Gender')
plt.ylabel('')
plt.show()
```

```python
df['Month'] = df['Date'].dt.to_period('M')

monthly_sales = df.groupby('Month')['Total_Amount'].sum()

monthly_sales.plot(kind='line', marker='o', color='green')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.grid()
plt.show()
```
